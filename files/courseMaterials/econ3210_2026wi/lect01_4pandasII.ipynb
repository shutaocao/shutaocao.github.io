### Missing values
The data have no missing, I'll create some.

Pandas uses the floating-point value NaN (Not a Number) to represent missing data.

```python
lfs.loc[lfs['emplPartBothSex']<1500,'emplPartBothSex']=np.nan
```

```python
lfs.head(100)
```

Methods of handling missing values

|Method	| Description |
|-------|-------------|
| dropna | Filter axis labels based on whether values for each label have missing data, with varying thresholds for how much missing data to tolerate.|
| fillna | Fill in missing data with some value or using an interpolation method such as "ffill" or "bfill".|
| isna | Return Boolean values indicating which values are missing/NA.|
| notna | Negation of isna, returns True for non-NA values and False for NA values.|

**Important**: These methods return new objects, and do not modify the content of original object.

```python
# Delete rows with missing values
lfs2 = lfs.loc[:,['ref_date','emplMale','emplPartBothSex','emplPartFemale']]
lfs2.dropna(how="all") # is a new object, but lfs2 is unmodified.
lfs2.head(20)
```

```python
lfs3 = lfs2.dropna()
lfs3.head(10)
```

```python
# if delete columns with missing values, do:
lfs3=lfs2.dropna(axis=1)
lfs3.head()
```

Fill in missing values with any thing you choose

```python
lfs3=lfs2.fillna(-1.0)
lfs3.head()
```

```python
## Can also use replace method
lfs3=lfs2.replace(np.nan, -2)
lfs3.head()
```

### Transforming Data Using a Function or Mapping
The `map` method is an iterable built-in function of Python. It iterates without needing to use `for` ... `in`.

The map method accepts a function or dictionary-like object containing a mapping to do the transformation of values:

```python
# create a new column based on column emplPartFemale
x = {832.4:-2, 833.9:-1}
lfs3['emplPartF'] = lfs3['emplPartFemale'].map(x)
lfs3
```

### Creating dummy variables
Dummy variables take values of 1 and 0. 

We can convert a categorical variable into a dummy matrix. For example, if a column in a DataFrame has 3 distinct values, we would derive a matrix or DataFrame with 3 columns containing 1s and 0s. This is done by function `get_dummies`.

We read in another data, create sex dummy variables, and merge them to main data.

```python
cansimid = '14100287'
filename = 'tbl14100287_bySex.csv'
#lfs = pd.read_csv(filename, index_col=[0,1])
lfs_bySex = pd.read_csv(filename,index_col=0)
lfs_bySex.head()
```

```python
dummies = pd.get_dummies(lfs_bySex["gender"], prefix="d_", dtype=float)
lfs_bySexDummy = lfs_bySex.join(dummies)
lfs_bySexDummy
```

## Time series data

This is another whole set of methods processing time series data.

I only show an example: set the LFS labor condition data to be time series, and convert monthly data to quarterly.

```python
# Convert the index to time series
import datetime
cansimid = '14100287'
filename = 'tbl14100287_bySex.csv'
lfs_bySex = pd.read_csv(filename,index_col=0)
lfs_bySex.index=pd.to_datetime(lfs_bySex.index)
# Convert monthly data to quarterly
lfsQtr = lfs_bySex.groupby('gender').resample('QS-OCT').mean()
lfsQtr = lfsQtr.round(2)
lfsQtr
```
