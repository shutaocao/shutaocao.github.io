## Functions
A Python function is a code block that implements a specific task and returns results, and can be used repeatedly.

Use `def` followed by function name to start the definition of function.

A function often ends with `return`.

A function may or may not have arguments.

Names defined and used in a function are local, i.e., they are not recognizable out of the function definition.
- Unless a variable is explicitly declared as a global variable.

````python
def say_hello(recipient):
    return f'Hello, {recipient}!'

economists = ['Feymann','Suzie','Victor']
for person in economists:
    print(say_hello(person))
````

````python
x = say_hello('Lucas')
x
````

````python
def powers(x):
    """
    Return a few powers of x.
    """
    y = [x ** 2, x ** 3, x ** 4]
    return y
````

````python
y = 2
z = powers(y)
print(z)
````

````python
def fib(n):    # write Fibonacci series up to n
    """Print a Fibonacci series up to n."""
    c, d = 0, 1
    while c < n:
        print(c, end=' ')
        c,d = d, c+d
    print()
# now call the functions
fib(500)
````

In the above, names (c,d,n) are local. they are undefined and not recognizable outside the function definition.

$n$ is called a function argument outside the function definition.

$n$ is called a parameter inside the function definition, meaning that we cannot assign another value to $n$ inside function definition.

````python
def fib2(n):
    """Return a list of Fibonacci series up to n."""
    a, b = 0, 1
    x = []
    while a < n:
        print(a, end=' ')
        x.append(a)
        a, b = b, a+b
    print()
    return x
# now call the functions
z = fib2(500)
print('z =',z)
````

````python
# calling function like this is also fine
z = fib2(n=500)
print(z)
````

One more example: solve $ax^2 + bx+c=0$ for $x$.

In defining the function, the default value for $c$ is $1$.

````python
import math
def roots(a, b, c=1):
    """ Return the roots of ax^2 + bx + c."""
    d = b**2 - 4*a*c
    r1 = (-b + math.sqrt(d)) / (2*a)
    r2 = (-b - math.sqrt(d)) / (2*a)
    return r1 , r2
````

````python
y = roots(2,5)
print(y)
````

````python
y = roots(2,5,1)
print(y)
````

````python
y = roots(2,5,3)
print(y)
````

````python
y = roots(b=5,c=0.5,a=2)
print(y)
````

### Variadic arguments: gather positional arguments with *
Python functions allow a variable number of positional arguments, using asterisk to group value of these arguments into a single _tuple_ of parameter values. 

These arguments are called **variadic arguments**, and they must be placed after other arguments in the function.

The function `print()` we have used is an example of functions with variadic arguments.

Example: two required arguments, x and y. The rest are optional and.

````python
def print_more(x, y, *args):
    print('x is required:', x)
    print('y is required too:', y)
    print('Other parameter values here:', args)
````

````python
print_more('B','A')
````

````python
print_more('A','B', 'C')
````

````python
print_more('A','B', 5, 'C', 4, '_', '?')
````

````python
x = [5, 'C', 4, '_', '?']
print_more('A','B', x)
````

````python
# To unpack the list (or a tuple) x, call the function using *x
x = [5, 'C', 4, '_', '?']
print_more('A','B', *x)
````

````python
# One more example
def product(first, *args):
    result = first
    for x in args:
        result = result * x
    return result
````

````python
product(2)
````

````python
product(2,3,4)
````

````python
# Note the output
x = [3,4]
product(2,x)
````

````python
# Note the output again!
x = [3,4]
product(2,*x)
````

### An example of functions in economics

````python
"""
this code solves for consumer's utility maximization problem. U(C,l) is Cobb-Douglas, l is leisure, C is consumption.
Budget constraint is C=wl + profit - tax.
I worked out first-order conditions in equations, translated them to code
"""
import numpy as np

# utility function
def utilityfun(l,C,shareC, sharel, sigma):
    tmplc = (l**sharel) * (C**shareC)
    return (tmplc**(1.0-sigma) - 1.0)/(1.0-sigma)

# optimal choices
def optimalChoice(sharel,shareC,wage,hmax,profit,tax):
    lstar = (sharel/(sharel+shareC)) *(wage*hmax + profit - tax) / wage
    Cstar = lstar * wage * shareC / sharel
    return lstar,Cstar

# parameters
shareC = 0.65
sharel = 1.0 - shareC
sigma = 1.4

wage, profit, tax = 0.5, 1.8, 0.5
hmax = 5.5
#---optimal choices: c/l = w * sharec/sharel
lstar, Cstar=optimalChoice(sharel,shareC,wage,hmax,profit,tax)
# maximized utility under wage
util0 = utilityfun(lstar,Cstar,shareC, sharel, sigma)

print('Optimal C=', Cstar, "optimal l=", lstar)
print('Maximized utility level=', util0)
````

### Objects

An object is a custom data structure containing both data (variables, called _attributes_) and code (functions, called _methods_).

Example: When we type num = 7, Python creates an object of type integer with the value 7, and assign an object reference to the name **num**.

### Classes
New objects are created using the `class` statement. A class typically consists of a collection
of functions that make up the methods. In Python a class can contain *attributes* (variables) and *methods* (functions).

The class definition usually contains a number of class method definitions (functions in a class).

* Each class method should have an argument `self` as its first argument. This object is a self-reference.

* Some class method names have special meaning, for example:
    * `__init__`: The name of the method that is invoked when the object is first created.
    * `__str__` : A method that is invoked when a simple string representation of the class is needed, as for example when printed.
    * There are more, see https://docs.python.org/3/reference/datamodel.html#special-method-names
 
### An example of classes

````python
%%file bank_account.py
# This class is drawn from books Learning Scietific Programming with Python and Python Distilled
class BankAccount:
    """An abstract base class representing a bank account"""
    num_Accounts = 0
    def __init__(self, customer, account_number, balance=0):
        """
        Initialize the BankAccount class with a customer, account number
        and opening balance (0 by default)
        """
        self.customer = customer
        self.account_number = account_number
        self.balance = balance
        BankAccount.num_Accounts += 1
    def __repr__(self):
        return f'BankAccount({self.customer!r},{self.account_number!r}, {self.balance!r})'
    def deposit(self, amount):
        """Amount deposited into the bank account"""
        if amount > 0:
            self.balance += amount
        else:
            print('Invalid deposit amount:', amount)
    def withdraw(self, amount):
        """Withdraw amount from the bank account"""
        if amount > 0:
            if amount > self.balance:
                print('Insufficient funds')
            else:
                self.balance -= amount
        else:
            print('Invalid withdrawal amount:', amount)
    def inquiry(self):
        return self.customer, self.account_number, self.balance
````

### Instances
An instance of a class is created by calling a class object as a function:

`object = ClassName(args)`

This creates a new instance that is then passed to the `__init__()` method. The arguments to `__init__()` method consists of the newly created instance _self_ along with the arguments supplied in calling the class object.

````python
from bank_account import BankAccount
# Example, create a few instances
account_a = BankAccount('Feyman', '202401081', 1000.0)
# The line above calls BankAccount.__init__(account_a, 'Feyman', '202401081', 1000.0)
account_b = BankAccount('Spock',  '20011236', 500.0)
# Calls BankAccount.__init__(account_b, 'Spock',  '20011236', 500.0)
````

### Decoding the example
`class BankAccount:`
- defines the class, names it as "BankAccount"

`num_Accounts = 0`
- Is a _class variable_, an attribute of the class as an object

`def __init__(self, customer, account_number, balance=0):`
- `__init_()` is a _special_ method of the class, that initializes an individual object from its class definition
- The argument _self_ specifies that it refers to the individual object itself, must come before other arguments
- In the instance `account_a = BankAccount('Feyman', '202401081', 1000.0)`, object "account_a" automatically passes as _self_, which means that _self_ is a reference to the instance "account_a" itself
- Every method call associated with the instance "account_a" automatically passes _self_, it gives the instance "account_a" access to the attributes and methods in the class "BankAccount"

`self.customer = customer`
- Takes the value associated with the parameter "customer" and assigns it to the variable (attribute) "customer", which is then attached to the instance "account_a" being created
- In `self.customer`, "customer" is an attribute, that is prefixed with _self_, by that, "customer" is available to every method in the class

The rest:
- The `__repr__()` method returns a string for viewing an object, it is optional
- Functions `deposit()`,`withdraw()`, and `inquiry()` are called instance _methods_
- An instance method is a function that operates on an instance of the class, which is passed as the first argument, _self_

## File Input and Output

Python can read text files from hard drives to memory, and can write data in memory to files.

### Reading files
Example: in hard drive, I have a file zen_Python.txt. We read it to memory using the Python built-in function `open()`.

````python
f = open('zen_Python.txt', mode='r', encoding="utf-8")
````

````python
f.<TAB>
````

The line above returns a corresponding file object _f_.

mode is an optional string that specifies the mode in which the file is opened.

The available modes are
* 'r', open file for reading (default).
* 'w', open file for writing; if the file does not exist, create a new file.
* 'x', open file for exclusive creation, failing if the file already exists
* 'a', open for writing, appending to the end of the existing file. If the file does not exist, it creates a new file. 
* 'b', open file in binary mode. 
* 'r+', open file (do not create) for reading and writing. 
* 'w+', open or create a file for writing and reading, discard existing contents. 
* 'a+', open or create file for reading and writing, and append data to end of file. 

````python
output = f.read()
f.close()
print(output)
type(output)
````

````python
# Reading a file with "with" will automatically close the file after reading.
with open('zen_Python.txt', mode='r', encoding="utf-8") as f:
    output = f.read()
print(output)
f.closed
````
