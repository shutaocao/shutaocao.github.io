The main [sequence types](https://docs.python.org/3/library/stdtypes.html#sequence-types-list-tuple-range) in Python are lists, tuples and range objects. Main differences between these sequence objects are:

* Lists are [mutable](https://docs.python.org/3/glossary.html#term-mutable) and their elements are usually *homogeneous* (things of the same type making a list of similar objects)

* Range objects are *efficient* sequences of integers (commonly used in `for` loops), use a small amount of memory and yield items only when needed

### Mutability

### List Copy and DeepCopy
Watch out the following: when the values of $a$ change, so do the values of $b$, without repeating operation `b = a`.

If we change the values of $b$, the values of $a$ will see the same change, even though we do not write to change $a$.

```python
a = [1,2,3,4,5]
b = 1*a
# b = a
print('b =',b)
a[2]=5.5
print('b =',b)
```

```python
a = [1,2,3,4,5]
b = a
print('b =',b)
b[2]=5.5
print('a =',a)
```

To create a new list that is independent of the list $a$, we should copy $a$ with one of the following methods
- `copy()`
- `list()`
- Slicing

```python
a = [1,2,3,4,5]
b = a.copy()
c = list(a)
d = a[:]
print('b = ',b)
print('c = ',c)
print('d = ',d)
```

Now changing the value of $a$ doe not affect values of $b,c,d$. 

```python
a[2]=5.5
print('a=',a)
print('b = ',b)
print('c = ',c)
print('d = ',d)
```

The method `.copy()` is _shallow_, it only works for the most outside list. For example: 

Below, $a$ is a list, one of its elements is also a list [4 ,5]. Using `.copy` or `list()` or slice does not make b,c,d independent of a.

```python
a = [1,2,3,[4,5]]
b = a.copy()
c = list(a)
d = a[:]
print('b = ',b)
print('c = ',c)
print('d = ',d)
```

```python
a[3][0]=9
print('b = ',b)
print('c = ',c)
print('d = ',d)
```

To make derived list fully independent of the original list, use method `deepcopy()`.

```python
import copy
a = [1,2,3,[4,5]]
b = copy.deepcopy(a)
print('b = ',b)
```

```python
a[3][0]=9
print('b = ',b)
```

```python
numbers = [1, 3.1, 3, 6, 5,9]
position = 0
while position < len(numbers):
    number = numbers[position]
    if number % 2 == 0:
        print('Found even number', number, 'position=',position)
        break
    position += 1
else: # break not called
    print('No even number found', 'position', position)
```

| `sum(sequence)` | return the sum of the entries of `sequence` |
| `max(sequence)`  | return the maximum value in `sequence` |
| `min(sequence)` | return the minimum value in `sequence` |
| `range(a,b,step)` | return the range object of integers from `a` to `b` (exclusive) by `step` |
| `list(sequence)` | return a list constructed from `sequence` |
| `sorted(sequence)` | return the sorted list from the items in `sequence` |
| `reversed(sequence)` | return the reversed iterator object from the items in `sequence` |
| `enumerate(sequence)` | return the enumerate object constructed from `sequence` |
| `zip(a,b)` | return an iterator that aggregates items from sequences `a` and `b` |

### Built-in Function Names

Do *not* use the names of these functions as variable names otherwise the reference to the built-in function will be lost. For example, do not use `sum`, `min`, `max`, `list` or `sorted` as a variable name. See the full list of [builtin functions](https://docs.python.org/3/library/functions.html).
