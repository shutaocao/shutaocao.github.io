## Pandas Overview

Pandas is a library in Python that is used to manipulate and analyze heterogeneous data. By now, it is the standard Python tool for data manipulation.

Pandas provides two objects:

- [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html): one-dimensional labeled array of any data type, integers, strings, Python objects etc.

- [DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame): a two-dimensional tabular data structure like a table with rows and columns. It is inspired by dataframe in R.

For (very) large data set (think of data with at least 1 billion rows or columns), one should use [https://dask.org](Dask), which we do not learn in this course.

Main features of pandas include
- data file input and output
- data wrangling: reshaping, slicing, aggregation, merging and joining, grouping etc.
- time series data: aggregation, etc.
- plotting

To use pandas, we need to import it. In using pandas, we often use Numpy modules.

The names np for numpy and pd for pandas are popular, so should be followed.

## DataFrame
The DataFrame object is a 2-dimensional labeled data structure with columns of potentially different types. We think of it like a spreadsheet table.

- A DataFrame can be created from:
    - Dict of 1D ndarrays, lists, dicts, or Series
    - 2-D numpy array
    - A list of dictionaries
    - Another Dataframe

Syntax:

``` python
df = pd.DataFrame(data, index = index, columns = columns)
```
- ```index```/ ``` columns ``` is a list of the row / column labels.

See [**pandas.DataFrame**](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.html) for a complete list of methods and classes on DataFrame.

#### DataFrame Creation 
Create data frame from dict of series or dictionaries:
- The index of the resulting DataFrame will be the union of the indices of the various Series. If there are any nested dictionaries, these will be first converted to Series.
- If no columns are passed, the columns will be the sorted list of dict keys.

````python
population_dict = {'California': 39538223, 'Texas': 29145505,
                   'Florida': 21538187, 'New York': 20201249,
                   'Pennsylvania': 13002700}
population = pd.Series(population_dict)
population
````

````python
area_dict = {'California': 423967, 'Texas': 695662, 'Florida': 170312, 
             'New York': 141297, 'Pennsylvania': 119280}
area = pd.Series(area_dict)
area
````

````python
states = pd.DataFrame({'population': population, 'area': area})
states
````

````python
states.index
````

````python
states.columns
````

````python
states.loc[:,'area']
````

Data frame can be constructed from series. 

In the above, we created a Series `population`, we make a data frame from it.

````python
pd.DataFrame(population, columns=['population'])
````

Create data frame from a list of dictionaries:

````python
data_dict = [{'a': i, 'b': 2 * i} for i in range(3)]
datax = pd.DataFrame(data_dict)
datax
````

Create data frame from Numpy arrays

````python
pd.DataFrame(np.random.rand(3, 2), columns=['foo', 'bar'],index=['a', 'b', 'c'])
````

````python
# Example: use pandas to create time series
import pandas as pd
np.random.seed(1234)
data = np.random.randn(10, 2)  # 5x2 matrix of N(0, 1) random draws
dates = pd.date_range(start='12/28/2010', periods=10, freq='W')
df = pd.DataFrame(data, columns=('price', 'weight'), index=dates)
print(df)
````

````python
df.mean()
````

Pandas can read in data from:

````python
# Example: read monthly Canadian labor market condition data in CSV form
import pandas as pd
cansimid = '14100287'
#filename = 'tbl14100287_wide.csv'
filename = 'tbl14100287Final3.csv'
lfs = pd.read_csv(filename, index_col=0)
````

````python
# Convert the index to time series
lfs.index = pd.to_datetime(lfs.index)
````

````python
lfs.index
````

````python
lfs.columns
````

Convert the monthly employment data to be quarterly:

````python
lfsQtr = lfs.resample('QS-OCT').mean()
lfsQtr = lfsQtr.round(2)
lfsQtr
````

Pandas can write data frames to files of format csv, excel, etc.

In the example below, lfsQtr is saved to the folder that holds the current .ipynb notebook.

We can specify the folder name to save lfsQtr to a designated folder.

````python
lfsQtr.to_csv('lfsQtr.csv')
````

We can also compress the csv file, useful if data size is large.

````python
compression_opts = dict(method='zip', archive_name='lfsQtr.csv')  
lfsQtr.to_csv('lfsQtr.csv.zip', index=True,compression=compression_opts)  
````

### Data Indexing and Selection
Two methods of indexing:
- The .loc[] attribute: indexing and slicing that references the explicit index (index labels)
- The .iloc[] attribute: indexing and slicing that references the implicit Python-style index, thus it is an indexing with integers.

Ways of slicing using .iloc are similar to those for Numpy arrays, Python lists and strings. 

````python
states
````

````python
states['California':'New York']
````

````python
states.loc['California':'New York']
````

````python
states.loc[['California','Florida'], ['area', 'population']]
````

````python
states.iloc[[0,2]]
````

````python
states.iloc[:-2]
````

Create a column

````python
states['density'] = states['population']/states['area']
states.head()
````

Selection with explicit rules

````python
states.loc[states.density > 100, ['population', 'density']]
````

Modify values

````python
states['party']='D'
states
````

````python
states.loc[['Texas','Florida'],['party']]='R'
states
````

````python
lfsQtr.loc['2019-01-01':'2020-04-01',['unemplRateBothSex']]
````

````python
lfsQtr.loc['2018-01-01':,['unemplRateBothSex']]
````
