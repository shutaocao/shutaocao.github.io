## Overview
[NumPy](https://numpy.org/) is the fundamental package for scientific computing in Python. 

It provides multidimensional array objects, various derived objects (such as masked arrays and matrices), and may routines of fast array operations.

These operations include
- mathematical and logical operations
- shape manipulation
- sorting
- discrete Fourier transforms
- basic linear algebra
- basic statistical operations
- random number generation.

### Differences between Numpy arrays and Python lists
- NumPy arrays have a fixed size at creation, while the length of a Python list can change dynamically.
- Elements in a NumPy array much have the same data type, while data types of elements in a Python list can be mixed.
- NumPy arrays allow mathematical and other operations on large numbers of data. Mathematical operations on list is very limited.
- Pretty much all scientific and mathematical Python-based packages use NumPy arrays, not Python lists.

## Array index and slicing

This table shows the slicing of one-dimensional arrays. They apply to one dimension in an n-dimensional array.
|  Slicing   |  Description   |
|  :-------  |  :--------  |
|  a[m]   |  select the $(m+1)$-th element  |
|  a[-m]   |  select the $m$th element from the end of array |
|  a[m:n]   |  select elements from position $m+1$ to position $n−1$ |
|  a[:]  |  select all elements in the given axis, same as a[0:-1]  |
|  a[:n]   |  select elements before the $n$th element |
|  a[m:]  |  select elements that follow the $(m+1)$-th element, same as a[m:-1] |
|  a[m:n:p]   |  select elements from $(m+1)$-th through the $(n-1)$th (exclusive), with increment p |
|  a[::-1]   |  select all the elements, in reverse order   |

### Reshaping
Reshaping an array take two types of order:
-  `order=‘C’` means to read / write the elements using C-like index order, with the last axis index changing fastest, back to the first axis index changing slowest. In the two-dimension case, `reshape` reads first the first row (which is the index 1), then the second row, and so on.
-  `order=‘F’` means to read / write the elements using Fortran-like index order, with the first index changing fastest, and the last index changing slowest. In the two-dimension case, `reshape` reads first the first column (which is the index 0), then the second column, and so on.

```python
a = np.fromfunction(lambda x, y: x*4+y+3, (3,4))
print('a =',a)
b = a.reshape((2,6))
print('b =',b)
```

```python
b = np.reshape(a,(2,6))  # does the same thing as b = a.reshape((2,6))
print('b =',b)
```

```python
b = a.reshape((2,6),order='C')
b
```

```python
b = a.reshape((2,6),order='F')
b
```

```python
b_c = a.reshape(12,order='C')
b_f = a.reshape(12,order='F')
print('b_c =', b_c)
print('b_f =', b_f)
```

Methods ravel and flatten appear to do the same thing. But according to Numpy manual:
- `ravel` returns a view of reshaped original array.
- `flatten` returns a copy of reshaped original array.

```python
a = np.fromfunction(lambda x, y: x*3+y*2+1, (3,4))
print('a =',a)
```

```python
b = a.ravel()
b
```

```python
b = a.ravel(order='F')
b
```

```python
print(b.shape)
b[1,]=-4
print('a =',a)
```

```python
b = a.flatten()
print(b.shape)
b
```

```python
b=a.flatten(order='F')
b
```

## Copies and Views
In operations and manipulations on an array, data in the array sometimes are copied to new arrays, but sometimes are not.

There are three cases.

Case 1: Simple assignments make no copy of objects or their data

```python
x = np.array([[0,1,2],[3,4,5]])
b = x
print(b)
```

```python
# b is x, changing elements in x also change b
x[1,2]=-5
print('x=',x)
print('b=',b)
```

Why is that? Recall, `x = np.array([[0,1,2],[3,4,5]])` creates an object (in this case 2 by 3 array of integers) with data (values) 0,1,2,3,4,5, and makes the name `x` point to the object together its values.

`b = x` is equivalent to saying "b is x", it means that Python makes the name `b` also point to the same object as the name `x` has pointed to.

`x[1,2]=-5` changes (the values of) the object. Since `b` also points to the same object, `b[1,2]` is now also changed to -5.

If we make change to the object through `b`, it will also change `x` since both `b` and `x` point to the same object.

```python
b[0,1] = 20
print('b=',b)
print('x=',x)
```

Case 2: The `copy` method makes a complete copy of an array and its data

```python
x = np.array([[0,1,2],[3,4,5]])
b = x.copy()
print(b)
```

Here, `b=x.copy()` means that Python created a new object (a 2 by 3 array of integers) that has the same data as the object that `x` represents.

`b` is independent of `x`, a change of the object through `b` does not affect the objected represented by `x`, vice versa.

`copy()` in Numpy is like the `deepcopy()` method in Python list.

```python
b[0,1]=20
print('b=',b)
print('x=',x)
```

Sometimes `copy` should be called after slicing if the original array is not required anymore.

```python
a = np.arange(int(1e8))        # a is a very large array
b = a[:20].copy()
del a
print(b)
```

Case 3: Different array objects can share the same data. Method `view` creates a new array object that _points to the same data_.

The method `view` in Numpy is like a shallow `copy` in Python.

```python
x = np.array([[0,1,2],[3,4,5]])
a = x.view()
print('a=',a)
# changing the shape of a does not change the shape of x
a=a.reshape((3,2))
print('a=',a)
print('x=',x)
# but changing the value of a also changes the value of x
a[1,1]=-10
print('a=',a)
print('x=',x)
```

### Arithmetic operations on more than one arrays
These operations are element wise.

- `x+y` is equivalent to `np.add(x,y)`
- `x-y` is equivalent to `np.subtract(x,y)`
- `x*y` is equivalent to `np.multiply(x,y)`
- `x / y` is equivalent to `np.divide(x,y)`

```python
x = np.random.normal(size=(2,3))
y = np.random.uniform(size=(2,3))
print('x =', np.round(x,decimals=4))
print('y =', np.round(y,decimals=4))
```

```python
# x*y is the element-wise product of two arrays, it is not the inner product.
x * y
```

```python
x+y
```

```python
x / y
```

### Matrix and vector operations
We focus on two-dimensional arrays, which we call matrices. We call one-dimensional arrays vectors.

matrix multiplication requires the dimensions to align.

```python
x = np.random.normal(size=(2,3))
y = np.random.uniform(size=(3,2))
print('x =', np.round(x,decimals=3))
print('y =', np.round(y,decimals=3))
```

```python
# matrix multiplication: use np.dot or @
np.dot(x,y)
```

```python
x @ y
```

```python
# We knew that x@y = y.T @ x.T
y.T @ x.T
```

```python
# Of course x@y!= y@x
y @ x
```

```python
# Inner product
np.inner(x[1,:],y[:,0])
```

```python
np.inner(x,y.T)               # = x@y
```

```python
# Diagonal
x = np.reshape(np.linspace(0,15,16), (4,4))
print('x =',x)
y=np.diag(x)
y
```
